Q1. Write a Python program to read and display a grayscale image. Display the following information:

- Image height and width
- Number of pixels
- Datatype
- Minimum pixel intensity
- Maximum pixel intensity
- Mean intensity
- Pixel value at a specified coordinate

Display the image using both OpenCV and Matplotlib.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "../input.jpg"
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully.")
```

```python
height, width = image.shape
print("Image Height:", height)
print("Image Width:", width)
```

```python
total_pixels = height * width
print("Number of Pixels:", total_pixels)
```

```python
datatype = image.dtype
print("Data Type:", datatype)
```

```python
min_intensity = np.min(image)
max_intensity = np.max(image)
mean_intensity = np.mean(image)

print("Minimum Pixel Intensity:", min_intensity)
print("Maximum Pixel Intensity:", max_intensity)
print("Mean Pixel Intensity:", mean_intensity)
```

```python
row = int(input("Enter the row no.: "))
column = int(input("Enter the column no.: "))

if 0 <= row < height and 0 <= column < width:
    pixel_value = image[row, column]
    print(f"Pixel value at ({row}, {column}):", pixel_value)
else:
    print("Specified coordinate is outside the image.")
```

```python
cv2.imshow("Grayscale Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.figure(figsize=(8, 6))
plt.title("Grayscale Image")
plt.imshow(image, cmap="gray")
plt.axis("off")
```
