Q1. Write a Python program to read and display a grayscale image. Display the following information:

    - Image height and width
    - Number of pixels
    - Datatype
    - Minimum pixel intensity
    - Maximum pixel intensity
    - Mean intensity
    - Pixel value at a specified coordinate

Display the image using both OpenCV and Matplotlib.

```python
# Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image in grayscale
image = cv2.imread('../input.jpg', cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError("Image not found. Check the image filename and path.")
```

```python
# Get image information
height, width = image.shape
num_pixels = image.size
data_type = image.dtype
min_intensity = np.min(image)
max_intensity = np.max(image)
mean_intensity = np.mean(image)

# Specify pixel coordinate (x, y)
x = 50
y = 50

print("Image Height:", height)
print("Image Width:", width)
print("Number of Pixels:", num_pixels)
print("Data Type:", data_type)
print("Minimum Pixel Intensity:", min_intensity)
print("Maximum Pixel Intensity:", max_intensity)
print("Mean Pixel Intensity:", mean_intensity)

if 0 <= x < width and 0 <= y < height:
    print(f"Pixel value at ({x}, {y}):", image[y, x])
else:
    print("The specified coordinate is outside the image.")
```

```python
# Display image using OpenCV
resized_image = cv2.resize(image, None, fx=0.2, fy=0.2)
cv2.imshow('Grayscale Image - OpenCV', resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Display image using Matplotlib
plt.imshow(image, cmap='gray')
plt.title('Grayscale Image - Matplotlib')
plt.axis('off')
plt.show()
```
