Q2. Perform the following pixel-level operations:

- Increase brightness
- Decrease brightness
- Increase contrast
- Decrease contrast
- Generate negative image
- Apply binary thresholding

Display the original image and all the resulting images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "../input.jpg"
image = cv2.imread(image_path)

if image is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully.")
```

```python
br_vl = 60
inc_br_img = cv2.add(image, br_vl)
dec_br_img = cv2.subtract(image, br_vl)
```

```python
alpha_inc = 1.7
alpha_dec = 0.6
inc_cn_img = cv2.convertScaleAbs(image, alpha=alpha_inc, beta=0)
dec_cn_img = cv2.convertScaleAbs(image, alpha=alpha_dec, beta=0)
```

```python
neg_img = cv2.subtract(255, image)
```

```python
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

threshold_value = 127

_, binary_img = cv2.threshold(
    gray_img,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)
```

```python
plt.figure(figsize=(15, 13))

plt.subplot(3, 3, 1)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(3, 3, 2)
plt.title("Increased Brightness")
plt.imshow(cv2.cvtColor(inc_br_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(3, 3, 3)
plt.title("Decreased Brightness")
plt.imshow(cv2.cvtColor(dec_br_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(3, 3, 4)
plt.title("Increased Contrast")
plt.imshow(cv2.cvtColor(inc_cn_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(3, 3, 5)
plt.title("Decreased Contrast")
plt.imshow(cv2.cvtColor(dec_cn_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(3, 3, 6)
plt.title("Negative Image")
plt.imshow(cv2.cvtColor(neg_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(3, 3, 7)
plt.title("Binary Image")
plt.imshow(binary_img, cmap="gray")
plt.axis("off")
```
