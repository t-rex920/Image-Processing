# Session 3 - Q2. Discrete Fourier Transform and Discrete Cosine Transform

Perform the following frequency-transform operations on a grayscale image:
1. Calculate the 2-D Discrete Fourier Transform (DFT).
2. Shift the zero-frequency component to the center.
3. Calculate and display the magnitude spectrum.
4. Apply logarithmic scaling to the spectrum.
5. Calculate the 2-D Discrete Cosine Transform (DCT).
6. Display the DCT coefficient matrix.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "input.jpg"
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError("Image could not be loaded.")
```

```python
# Calculate 2-D DFT
dft = cv2.dft(np.float32(image), flags=cv2.DFT_COMPLEX_OUTPUT)

# Shift zero-frequency component to the center
dft_shift = np.fft.fftshift(dft)

# Calculate magnitude spectrum
magnitude = cv2.magnitude(dft_shift[:, :, 0], dft_shift[:, :, 1])

# Apply logarithmic scaling
log_magnitude = np.log1p(magnitude)
```

```python
# Display DFT results
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(magnitude, cmap="gray")
plt.title("Magnitude Spectrum")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(log_magnitude, cmap="gray")
plt.title("Log Magnitude Spectrum")
plt.axis("off")
```

```python
# Calculate 2-D DCT
dct = cv2.dct(np.float32(image))

# Display DCT coefficient matrix
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(np.log1p(np.abs(dct)), cmap="gray")
plt.title("DCT Coefficient Matrix")
plt.axis("off")
```
