Q1. Write a Python program to perform the following geometric transformations on an image:

- Translation
- Scaling
- Rotation
- Horizontal reflection
- Vertical reflection

Display the original and transformed images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "../input.jpg"
image = cv2.imread(image_path)

if image is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully.")

height, width = image.shape[0 : 2]
print("Height:", height)
print("Width:", width)
```

```python
tr_x = 100
tr_y = 50

M_tr = np.float32([
    [1, 0, tr_x],
    [0, 1, tr_y]
])
M_tr.shape

translated_img = cv2.warpAffine(
    image,
    M_tr,
    (image.shape[1], image.shape[0])
)
```

```python
scaled_img = cv2.resize(
    image,
    None,
    fx=1.5,
    fy=1.5,
    interpolation=cv2.INTER_LINEAR
)
```

```python
M_rotation = cv2.getRotationMatrix2D(
    (width // 2, height // 2),
    45,
    1
)

rotated_img = cv2.warpAffine(image, M_rotation, (width, height))
```

```python
horizontal_img = cv2.flip(image, 1)
vertical_img = cv2.flip(image, 0)
```

```python
plt.figure(figsize=(15, 13))

plt.subplot(2, 3, 1)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 3, 2)
plt.title("Translated Image")
plt.imshow(cv2.cvtColor(translated_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 3, 3)
plt.title("Scaled Image")
plt.imshow(cv2.cvtColor(scaled_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 3, 4)
plt.title("Rotated Image")
plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 3, 5)
plt.title("Horizontal Image")
plt.imshow(cv2.cvtColor(horizontal_img, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 3, 6)
plt.title("Vertical Image")
plt.imshow(cv2.cvtColor(vertical_img, cv2.COLOR_BGR2RGB))
plt.axis("off")
```
