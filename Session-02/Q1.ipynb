Q1. Write a Python program to perform the following geometric transformations on an image:

    - Translation
    - Scaling
    - Rotation
    - Horizontal reflection
    - Vertical reflection

Display the original and transformed images.

```python
# Import libraries
import cv2
import matplotlib.pyplot as plt
```

```python
# Read the image
image = cv2.imread('../input.jpg', cv2.IMREAD_COLOR)

if image is None:
    raise FileNotFoundError("Image not found. Check the image filename and path.")

height, width = image.shape[:2]
```

```python
# a. Translation
tx = 50
ty = 30
translation_matrix = cv2.getRotationMatrix2D((0, 0), 0, 1)
translation_matrix[0, 2] = tx
translation_matrix[1, 2] = ty
translation = cv2.warpAffine(image, translation_matrix, (width, height))

# b. Scaling
scaling = cv2.resize(image, None, fx=1.5, fy=1.5)

# c. Rotation
center = (width // 2, height // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1)
rotation = cv2.warpAffine(image, rotation_matrix, (width, height))

# d. Horizontal reflection
horizontal = cv2.flip(image, 1)

# e. Vertical reflection
vertical = cv2.flip(image, 0)
```

```python
# Display original and transformed images using Matplotlib
titles = ['Original', 'Translation', 'Scaling', 'Rotation',
          'Horizontal Reflection', 'Vertical Reflection']
images = [image, translation, scaling, rotation, horizontal, vertical]

plt.figure(figsize=(15, 10))  # Increased width for 6 subplots

for i in range(len(images)):
    plt.subplot(2, 3, i + 1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))

    # Optional: Display current dimensions in the title to show it scaled
    h, w = images[i].shape[:2]
    plt.title(f"{titles[i]}\n({w}x{h})")

    plt.axis("off")

plt.tight_layout()
plt.show()
```
