Q2. Perform image zooming and shrinking using different interpolation techniques:

- Nearest-neighbor interpolation
- Bilinear interpolation
- Bicubic interpolation

Perform:
- 2× zooming
- 4× zooming
- 1/2 shrinking
- 1/4 shrinking

Compare the visual quality of the resulting images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "../input.jpg"
image = cv2.imread(image_path)

if image is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully.")

height, width = image.shape[0:2]
print("Original Height:", height)
print("Original Width:", width)
```

```python
nearest_2x = cv2.resize(image, None, fx=2, fy=2, interpolation=cv2.INTER_NEAREST)
nearest_4x = cv2.resize(image, None, fx=4, fy=4, interpolation=cv2.INTER_NEAREST)
nearest_half = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_NEAREST)
nearest_quarter = cv2.resize(image, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_NEAREST)
```

```python
bilinear_2x = cv2.resize(image, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)
bilinear_4x = cv2.resize(image, None, fx=4, fy=4, interpolation=cv2.INTER_LINEAR)
bilinear_half = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)
bilinear_quarter = cv2.resize(image, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_LINEAR)
```

```python
bicubic_2x = cv2.resize(image, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
bicubic_4x = cv2.resize(image, None, fx=4, fy=4, interpolation=cv2.INTER_CUBIC)
bicubic_half = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_CUBIC)
bicubic_quarter = cv2.resize(image, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_CUBIC)
```

```python
plt.figure(figsize=(7, 5))
plt.title("Original Image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")
```

```python
plt.figure(figsize=(16, 12))

plt.subplot(2, 2, 1)
plt.title("Nearest - 2x")
plt.imshow(cv2.cvtColor(nearest_2x, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 2)
plt.title("Nearest - 4x")
plt.imshow(cv2.cvtColor(nearest_4x, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 3)
plt.title("Nearest - 1/2")
plt.imshow(cv2.cvtColor(nearest_half, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 4)
plt.title("Nearest - 1/4")
plt.imshow(cv2.cvtColor(nearest_quarter, cv2.COLOR_BGR2RGB))
plt.axis("off")
```

```python
plt.figure(figsize=(16, 12))

plt.subplot(2, 2, 1)
plt.title("Bilinear - 2x")
plt.imshow(cv2.cvtColor(bilinear_2x, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 2)
plt.title("Bilinear - 4x")
plt.imshow(cv2.cvtColor(bilinear_4x, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 3)
plt.title("Bilinear - 1/2")
plt.imshow(cv2.cvtColor(bilinear_half, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 4)
plt.title("Bilinear - 1/4")
plt.imshow(cv2.cvtColor(bilinear_quarter, cv2.COLOR_BGR2RGB))
plt.axis("off")
```

```python
plt.figure(figsize=(16, 12))

plt.subplot(2, 2, 1)
plt.title("Bicubic - 2x")
plt.imshow(cv2.cvtColor(bicubic_2x, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 2)
plt.title("Bicubic - 4x")
plt.imshow(cv2.cvtColor(bicubic_4x, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 3)
plt.title("Bicubic - 1/2")
plt.imshow(cv2.cvtColor(bicubic_half, cv2.COLOR_BGR2RGB))
plt.axis("off")

plt.subplot(2, 2, 4)
plt.title("Bicubic - 1/4")
plt.imshow(cv2.cvtColor(bicubic_quarter, cv2.COLOR_BGR2RGB))
plt.axis("off")
```
