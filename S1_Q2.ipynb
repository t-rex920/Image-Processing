# Session 1 - Q2. Pixel-Level Image Operations

Perform the following pixel-level operations on a grayscale image:
1. Increase brightness
2. Decrease brightness
3. Increase contrast
4. Decrease contrast
5. Generate negative image
6. Apply binary thresholding

Display the original image and all resulting images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "input.jpg"
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError("Image could not be loaded.")
```

```python
brightness_value = 50
increase_brightness = cv2.add(image, brightness_value)
decrease_brightness = cv2.subtract(image, brightness_value)

increase_contrast = cv2.convertScaleAbs(image, alpha=1.5, beta=0)
decrease_contrast = cv2.convertScaleAbs(image, alpha=0.5, beta=0)

negative_image = 255 - image

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

threshold_value = 127
_, binary_image = cv2.threshold(
    gray_image,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)
```

```python
images = [
    image,
    increase_brightness,
    decrease_brightness,
    increase_contrast,
    decrease_contrast,
    negative_image,
    binary_image
]

titles = [
    "Original", "Increase Brightness", "Decrease Brightness",
    "Increase Contrast", "Decrease Contrast", "Negative",
    "Binary Thresholding"
]

plt.figure(figsize=(14, 10))

for i in range(len(images)):
    plt.subplot(3, 3, i + 1)

    if i == 6:
        plt.imshow(images[i], cmap="gray")
    else:
        plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))

    plt.title(titles[i])
    plt.axis("off")
```
