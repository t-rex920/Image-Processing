# Session 2 - Q2. Image Zooming and Shrinking

Perform image zooming and shrinking using different interpolation techniques:
1. Nearest-neighbor interpolation
2. Bilinear interpolation
3. Bicubic interpolation

Perform 2x zooming, 4x zooming, 1/2 shrinking and 1/4 shrinking. Compare the visual quality of the resulting images.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "input.jpg"
image = cv2.imread(image_path)

if image is None:
    raise FileNotFoundError("Image could not be loaded. Check input.jpg.")

height, width = image.shape[:2]
```

```python
# 2x zooming
nearest_2x = cv2.resize(image, (width * 2, height * 2), interpolation=cv2.INTER_NEAREST)
bilinear_2x = cv2.resize(image, (width * 2, height * 2), interpolation=cv2.INTER_LINEAR)
bicubic_2x = cv2.resize(image, (width * 2, height * 2), interpolation=cv2.INTER_CUBIC)

# 4x zooming
nearest_4x = cv2.resize(image, (width * 4, height * 4), interpolation=cv2.INTER_NEAREST)
bilinear_4x = cv2.resize(image, (width * 4, height * 4), interpolation=cv2.INTER_LINEAR)
bicubic_4x = cv2.resize(image, (width * 4, height * 4), interpolation=cv2.INTER_CUBIC)

# 1/2 shrinking
nearest_half = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_NEAREST)
bilinear_half = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)
bicubic_half = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_CUBIC)

# 1/4 shrinking
nearest_quarter = cv2.resize(image, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_NEAREST)
bilinear_quarter = cv2.resize(image, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_LINEAR)
bicubic_quarter = cv2.resize(image, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_CUBIC)
```

```python
images = [
    image, nearest_2x, bilinear_2x, bicubic_2x,
    nearest_4x, bilinear_4x, bicubic_4x,
    nearest_half, bilinear_half, bicubic_half,
    nearest_quarter, bilinear_quarter, bicubic_quarter
]
titles = [
    "Original", "Nearest 2x", "Bilinear 2x", "Bicubic 2x",
    "Nearest 4x", "Bilinear 4x", "Bicubic 4x",
    "Nearest 1/2", "Bilinear 1/2", "Bicubic 1/2",
    "Nearest 1/4", "Bilinear 1/4", "Bicubic 1/4"
]

plt.figure(figsize=(16, 16))

for i in range(len(images)):
    plt.subplot(4, 4, i + 1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis("off")
```
